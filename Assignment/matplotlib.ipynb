```python
days = ["Monday","Tuesday","Wednesday","Thursday","Friday"]
temperature_1 = [30,32,35,33,36]
temperature_2 = [28,31,34,32,35]

plt.figure(figsize=(8,4))

plt.plot(days, temperature_1, marker='o', label="Temp 1")
plt.plot(days, temperature_2, marker='s', label="Temp 2")

plt.grid(True)

plt.title("Weekly Temperature Comparison")

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.legend()

plt.show()
```

```python
subjects = ["Maths","Science","English"]
marks_2024 = [80,75,90]
marks_2025 = [85,80,95]
x = np.arange(len(subjects))
width = 0.35

plt.figure(figsize = (8,4))

plt.bar(x - width/2,marks_2024,width,label = "2024",color = "crimson")
plt.bar(x + width/2,marks_2025,width,label = "2025")

plt.title("subject-wise marks comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(x,subjects)
plt.legend()

plt.show()
```

```python
products = ["Laptop","Phone","Tablet"]
online = [50,70,40]
offline = [30,40,20]

plt.figure(figsize=(8,4))

plt.bar(products, online, label="Online",color = "pink")
plt.bar(products, offline, label="Offline", bottom=online,color = "lightgreen")

plt.title("Stacked Sales Report")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.legend()
plt.show()
```
