Q24. Count missing values in each column.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Rahul", "Sara", "Vikram"],
    "Age": [18, np.nan, 18, 20, 19],
    "Score": [85, 92, np.nan, 88, 90]
}

df = pd.DataFrame(data)

print(df.isnull().sum())
```

Q25. Fill missing values with a fixed value using fillna().

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Rahul", "Sara", "Vikram"],
    "Age": [18, np.nan, 18, 20, 19],
    "Score": [85, 92, np.nan, 88, 90]
}

df = pd.DataFrame(data)

df_filled = df.fillna(0)

print(df_filled)
```

Q26. Drop rows containing missing values using dropna().

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Rahul", "Sara", "Vikram"],
    "Age": [18, np.nan, 18, 20, 19],
    "Score": [85, 92, np.nan, 88, 90]
}

df = pd.DataFrame(data)

df_clean = df.dropna()

print(df_clean)
```

Q27. Create a new column Total = Price * Quantity.

```python
import pandas as pd

data = {
    "Product": ["Pen", "Pencil", "Notebook"],
    "Price": [10, 5, 50],
    "Quantity": [2, 4, 3]
}

df = pd.DataFrame(data)

df["Total"] = df["Price"] * df["Quantity"]

print(df)
```

Q28. Rename a column using rename().

```python
data = {
    "Name": ["Amit", "Neha", "Rahul"],
    "Score": [85, 92, 78]
}

df = pd.DataFrame(data)

# Rename column "Score" to "Marks"
df = df.rename(columns={"Score": "Marks"})

print(df)
```
