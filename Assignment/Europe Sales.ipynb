```python
sales = sales.drop_duplicates()

sales.shape
```

```python
country_revenue = sales.groupby("Country")["Total Revenue"].sum()
country_revenue.sort_values(
    ascending=False
).head(10)
```

```python
country_profit = sales.groupby("Country")["Total Profit"].sum()

country_profit
```

```python
country_profit.sort_values().head(5)
```

```python
item_units = sales.groupby(
    "Item Type"
)["Units Sold"].sum()

item_units
```

```python
item_revenue = sales.groupby(
    "Item Type"
)["Total Revenue"].sum()

item_revenue.idxmax()
```

```python
item_profit = sales.groupby(
    "Item Type"
)["Total Profit"].sum()

item_profit.idxmax()
```

```python
channel_revenue = sales.groupby(
    "Sales Channel"
)["Total Revenue"].sum()

channel_revenue
```

```python
channel_profit = sales.groupby(
    "Sales Channel"
)["Total Profit"].sum()

channel_profit
```

```python
channel_profit.idxmax()
```

```python
sales["Order Priority"].value_counts()
```

```python
avg_profit = sales.groupby(
    "Order Priority"
)["Total Profit"].mean()

avg_profit
```

```python
avg_profit.idxmax()
```

```python
sales["Total Revenue"].corr(
    sales["Total Profit"]
)
```

```python
correlation = sales["Total Revenue"].corr(
    sales["Total Profit"]
)

if correlation > 0:
    print("Higher revenue generally leads to higher profit")
else:
    print("No positive relationship")
```

```python
sales["Order Date"] = pd.to_datetime(
    sales["Order Date"]
)
```

```python
sales["Month"] = sales["Order Date"].dt.month_name()
```

```python
monthly_sales = sales.groupby(
    "Month"
)["Total Revenue"].sum()

monthly_sales
```

```python
monthly_sales.idxmax()
```

```python
monthly_sales.idxmin()
```

```python
summary = {
    "Total Revenue": sales["Total Revenue"].sum(),
    "Total Profit": sales["Total Profit"].sum(),
    "Best Country": country_revenue.idxmax(),
    "Best Item Type": item_units.idxmax(),
    "Best Sales Channel": channel_profit.idxmax(),
    "Total Orders": sales["Order ID"].count()
}

summary
```
