```python
# Null values in columns
null_col = movies.isnull().sum()
print(null_col)

# Null values in rows
null_row = movies.isnull().sum(axis=1)
print(null_row)

# Percentage null values
null_percent = round((movies.isnull().sum()/len(movies))*100,2)
print(null_percent)
```

```python
drop_cols = [
    'color',
    'director_facebook_likes',
    'actor_1_facebook_likes',
    'actor_2_facebook_likes',
    'actor_3_facebook_likes',
    'actor_2_name',
    'cast_total_facebook_likes',
    'actor_3_name',
    'duration',
    'facenumber_in_poster',
    'content_rating',
    'country',
    'movie_imdb_link',
    'aspect_ratio',
    'plot_keywords'
]

movies = movies.drop(columns=drop_cols)
```

```python
# After dropping unnecessary columns
null_percent = round((movies.isnull().sum()/len(movies))*100,2)

high_null_cols = null_percent[null_percent > 5].index

movies = movies.dropna(subset=high_null_cols)
```

```python
movies = movies[movies.isnull().sum(axis=1) <= 5]
```

```python
movies['language'].fillna('English', inplace=True)
```

```python
retained_rows = len(movies)

original_rows = pd.read_csv("Movie-Assignment-Data-1.csv").shape[0]

retained_percent = round((retained_rows/original_rows)*100,2)

print("Rows Retained:", retained_rows)
print("Retention %:", retained_percent)
```

```python
movies['budget'] = movies['budget']/1000000
movies['gross'] = movies['gross']/1000000
movies
```

```python
movies['profit'] = movies['gross'] - movies['budget']

top10 = (
    movies.sort_values(by='profit', ascending=False)
          .head(10)
)

top10[['movie_title','profit']]
```

```python
movies = movies.drop_duplicates()

movies['profit'] = movies['gross'] - movies['budget']

top10 = (
    movies.sort_values(by='profit', ascending=False)
          .head(10)
)

top10[['movie_title','profit']]
```

```python
IMDb_Top_250 = (
    movies[movies['num_voted_users'] > 25000]
    .sort_values(by='imdb_score', ascending=False)
    .head(250)
    .copy()
)

IMDb_Top_250['Rank'] = range(1,251)

IMDb_Top_250.head()
```

```python
Top_Foreign_Lang_Film = IMDb_Top_250[
    IMDb_Top_250['language'] != 'English'
]

Top_Foreign_Lang_Film[['movie_title','language']]
```

```python
top10director = (
    movies.groupby('director_name')['imdb_score']
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top10director
```

```python
genres_split = movies['genres'].str.split('|')

movies['genre_1'] = genres_split.str[0]
movies['genre_2'] = genres_split.str[1].fillna(genres_split.str[0])
```

```python
PopGenre = (
    movies.groupby(['genre_1','genre_2'])['gross']
    .mean()
    .sort_values(ascending=False)
    .head(5)
    .reset_index()
)

PopGenre
```

```python
Meryl_Streep = movies[
    movies['actor_1_name'] == 'Meryl Streep'
]

Leo_Caprio = movies[
    movies['actor_1_name'] == 'Leonardo DiCaprio'
]

Brad_Pitt = movies[
    movies['actor_1_name'] == 'Brad Pitt'
]
```

```python
Combined = pd.concat(
    [Meryl_Streep, Leo_Caprio, Brad_Pitt],
    ignore_index=True
)
```

```python
actor_stats = (
    Combined.groupby('actor_1_name')
    [['num_critic_for_reviews',
      'num_user_for_reviews']]
    .mean()
)

print(actor_stats)
```

```python
critic_favorite = actor_stats['num_critic_for_reviews'].idxmax()

print("Critic Favorite Actor:", critic_favorite)
```

```python
audience_favorite = actor_stats['num_user_for_reviews'].idxmax()

print("Audience Favorite Actor:", audience_favorite)
```
